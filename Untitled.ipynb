```python
## 1. 导入 PDF 并解析
from knowledge_base.pdf_parser import PDFParser

pdf_path = "data/documents/毛概大题整理.pdf"

pages = PDFParser.parse(pdf_path)
print(f"共解析出 {len(pages)} 页文本\n")

for p in pages:
    print(f"--- 第{p['page']+1}页 (字数: {len(p['text'])}) ---")
    print(p['text'][:200])
    print("...\n")
```

```python
## 2. 文本切片
from knowledge_base.text_splitter import TextSplitter

splitter = TextSplitter(chunk_size=500, chunk_overlap=50)
chunks = splitter.split_pages(pages)

print(f"共切分为 {len(chunks)} 个文本块\n")
for i, chunk in enumerate(chunks[:5]):
    print(f"--- Chunk {i} (字数: {len(chunk['text'])}) ---")
    print(f"来源: {chunk['metadata']['filename']} 第{chunk['metadata']['page']+1}页")
    print(chunk['text'][:200])
    print("...\n")
```

```python
## 3. 向量存储 & 索引
from knowledge_base.vector_store import VectorStore

vector_store = VectorStore(
    collection_name="maogai_knowledge",
    persist_dir="data/chroma_db",
)

# 清空旧数据后重新索引
vector_store.clear()
count = vector_store.add_chunks(chunks)
print(f"已索引 {count} 个文本块")
print(f"向量维度: {len(vector_store.model.encode(['测试'])[0])}")
```

```python
## 4. 检索测试
results = vector_store.search("毛泽东思想的形成过程", top_k=3)

for i, r in enumerate(results):
    print(f"\n=== 结果 {i+1} (相似度: {r['score']:.4f}) ===")
    print(f"来源: {r['metadata']['filename']} 第{r['metadata']['page']+1}页")
    print(f"内容: {r['text'][:300]}...")
```
